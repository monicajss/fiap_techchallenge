# Exploração dos Tipos de Dados: Amazon

Notebook inicial para conhecer a estrutura da base de produtos e avaliações da Amazon. O foco é identificar os tipos inferidos pelo pandas, valores ausentes e características que serão importantes na preparação dos dados. A base original não será alterada.

## 1. Importar Bibliotecas Necessárias

Usaremos pandas para leitura e exploração tabular, e `Path` para localizar o CSV de forma independente do diretório de execução.

```python
from pathlib import Path

import pandas as pd
from IPython.display import display

# Configurações para melhor visualização do DataFrame
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 100)
```

## 2. Carregar a Base de Dados da Amazon

O arquivo esperado é `data/raw/amazon.csv`. O bloco abaixo procura o CSV na raiz do projeto e também um nível acima, caso o notebook seja executado a partir da pasta `notebooks`.

```python
# Carregando o dataset
candidate_paths = [
    Path("data/raw/amazon.csv"),
    Path("../data/raw/amazon.csv"),
]
data_path = next((path.resolve() for path in candidate_paths if path.exists()), None)

# Verificando se o arquivo foi encontrado
if data_path is None:
    raise FileNotFoundError(
        "Não encontrei data/raw/amazon.csv. Verifique se o dataset está na pasta esperada."
    )

# Carregando o DataFrame
df = pd.read_csv(data_path)

# Exibindo informações sobre o DataFrame
print(f"Arquivo: {data_path}")
print(f"Dimensões: {df.shape[0]:,} linhas x {df.shape[1]} colunas")
```

## 3. Visualizar as Primeiras Linhas dos Dados

A amostra ajuda a entender o formato dos campos antes de interpretar os tipos inferidos automaticamente.

```python
# Exibindo as primeiras linhas do DataFrame
display(df.head())
```

## 4. Identificar os Tipos de Dados das Colunas

`dtypes` mostra como o pandas leu cada campo. Valores com `object` podem ser texto de fato ou números armazenados como texto, por exemplo preços com símbolo de moeda e avaliações.

```python
# Exibindo os tipos inferidos pelo pandas
display(                                # Exibe os tipos de dados inferidos pelo pandas para cada coluna do DataFrame
    df.dtypes.rename("tipo_pandas")
    .to_frame()
    .assign(coluna=lambda tabela: tabela.index)
    [["coluna", "tipo_pandas"]]
    .reset_index(drop=True)
)

# Exibindo a distribuição dos tipos inferidos
print("\nDistribuição dos tipos inferidos:")
display(df.dtypes.astype(str).value_counts().rename_axis("tipo").to_frame("quantidade_de_colunas")) # Exibe a contagem de colunas para cada tipo inferido
```

## 5. Analisar Valores Únicos e Ausentes

A tabela resume o total de valores ausentes e distintos por coluna. A proporção de ausentes facilita comparar campos com quantidades diferentes de registros.

```python
# Analisando a qualidade dos dados
quality_summary = pd.DataFrame(
    {
        "valores_ausentes": df.isna().sum(),   # Calcula o número de valores ausentes em cada coluna
        "percentual_ausente": (df.isna().mean() * 100).round(2), # Calcula o percentual de valores ausentes em cada coluna
        "valores_unicos": df.nunique(dropna=True), # Calcula o número de valores únicos em cada coluna
    }
).sort_values(["valores_ausentes", "valores_unicos"], ascending=[False, True]) # Ordena o resumo da qualidade dos dados pelas colunas com mais valores ausentes e em caso de empate, pelas colunas com menos valores únicos.

# Exibindo o resumo da qualidade dos dados
display(quality_summary)
```

## 6. Resumir Estatísticas dos Dados Numéricos e Categóricos

`describe()` gera estatísticas para campos numéricos e um resumo separado para campos de texto. Em seguida, `value_counts()` mostra os valores mais frequentes em categoria e avaliação; a avaliação pode estar como texto no CSV original.

```python
# Identificando colunas numéricas e de texto
numeric_columns = df.select_dtypes(include="number").columns    # Seleciona colunas com tipo numérico
text_columns = df.select_dtypes(include="object").columns       # Seleciona colunas com tipo object

# Exibindo informações sobre colunas numéricas e de texto
print(f"Colunas numéricas reconhecidas pelo pandas: {len(numeric_columns)}")    # Exibe a quantidade de colunas numéricas reconhecidas pelo pandas
if len(numeric_columns) > 0:
    display(df[numeric_columns].describe().T)
else:
    print("Nenhuma coluna numérica foi reconhecida; valores numéricos podem estar armazenados como texto.")
    
# Exibindo informações sobre colunas de texto
print(f"Colunas de texto/object: {len(text_columns)}")
if len(text_columns) > 0:
    display(df[text_columns].describe().T)
```

```python
# Exibindo os valores mais frequentes para colunas específicas
for column in ["category", "rating"]:
    if column in df.columns:
        print(f"\nValores mais frequentes em {column}:")
        display(df[column].value_counts(dropna=False).head(10).to_frame("frequencia"))
```

### Percepção inicial

No Pandas, o tipo `object` geralmente indica texto. Antes de analisar os dados, é importante verificar os campos de preço, desconto, avaliação e quantidade. Algumas colunas podem conter vários valores separados por vírgula, sendo necessário entender como estão organizados antes de contar avaliações ou usuários.

```python
# Diagnóstico antes de definir as regras de tratamento
numeric_candidates = [
    "discounted_price",
    "actual_price",
    "discount_percentage",
    "rating",
    "rating_count",
]

# Verificando se as colunas candidatas a numéricas podem ser convertidas para tipo numérico
for column in numeric_candidates:
    cleaned_values = (
        df[column]
        .astype("string")
        .str.replace(r"[₹,%\s]", "", regex=True)
    )
    parsed_values = pd.to_numeric(cleaned_values, errors="coerce")
    invalid_mask = df[column].notna() & parsed_values.isna()
    print(f"{column}: {invalid_mask.sum()} valores não convertidos")
    if invalid_mask.any():
        display(df.loc[invalid_mask, [column]].drop_duplicates().head(10))

print("\nNíveis da hierarquia de category:")
category_level_counts = df["category"].str.split("|", regex=False).str.len()
display(category_level_counts.value_counts().sort_index().rename("quantidade"))

print("\nAmostras dos campos de reviews agrupados:")
display(df[["user_id", "review_id", "review_title", "review_content"]].head(3))
```

```python
# Analisando a consistência das colunas de reviews agrupados
review_list_columns = ["user_id", "review_id", "review_title", "review_content"]
review_item_counts = pd.DataFrame(
    {
        column: df[column].fillna("").str.count(",").add(1).where(df[column].notna())
        for column in review_list_columns
    }
)

# Exibindo a quantidade de itens separados por vírgula nos campos agrupados
print("Quantidade de itens separados por vírgula nos campos agrupados:")
display(review_item_counts.describe().T[["mean", "min", "max"]].round(2)) # Exibe a média, mínimo e máximo da quantidade de itens separados por vírgula nos campos agrupados, considerando apenas as linhas não nulas.
print("\nProporção de linhas em que as contagens diferem de review_id:")
display(        # Exibe a proporção de linhas em que as contagens de itens separados por vírgula diferem da contagem de review_id, calculando a média das comparações booleanas e convertendo para percentual
    review_item_counts.ne(review_item_counts["review_id"], axis=0)
    .mean()
    .mul(100)
    .round(2)
    .rename("percentual_de_linhas")
    .to_frame()
)
```

Os campos de usuário e `review_id` não serão exportados; títulos e comentários permanecem agrupados.

## 8. Tratar os Dados para o Power BI

A transformação será feita em `df_powerbi`, uma cópia da base. Preços ficam numéricos em INR (Rúpia Indiana), desconto vira fração (por exemplo, `0.64` para 64%), `rating` inválido e ausências permanecem nulos, e o caminho de categoria ganha colunas por nível. Nenhuma linha será removida; `record_id` identifica cada linha original. Reviews permanecem agregados porque a separação por vírgula não é confiável.

```python
# Criando a cópia analítica e uma chave única para cada linha da fonte
df_powerbi = df.copy()
df_powerbi.insert(0, "record_id", range(1, len(df_powerbi) + 1))

# Limpando espaços e campos de texto vazios
string_columns = df_powerbi.select_dtypes(include="object").columns
for column in string_columns:
    df_powerbi[column] = df_powerbi[column].astype("string").str.strip()
    df_powerbi[column] = df_powerbi[column].replace("", pd.NA)

# Convertendo campos numéricos que vieram como texto
for column in ["discounted_price", "actual_price", "discount_percentage", "rating", "rating_count"]:
    numeric_text = (
        df_powerbi[column]
        .astype("string")
        .str.replace(r"[₹,%\s]", "", regex=True)
    )
    df_powerbi[column] = pd.to_numeric(numeric_text, errors="coerce")

# Convertendo o desconto para fração percentual e a contagem para inteiro anulável
df_powerbi["discount_percentage"] = df_powerbi["discount_percentage"] / 100
df_powerbi["rating_count"] = df_powerbi["rating_count"].astype("Int64")

# Separando o caminho hierárquico em níveis de categoria
category_parts = df_powerbi["category"].str.split("|", regex=False, expand=True)
for level in range(category_parts.shape[1]):
    level_column = f"category_level_{level + 1}"
    df_powerbi[level_column] = category_parts[level].astype("string").str.strip()
    df_powerbi[level_column] = df_powerbi[level_column].replace("", pd.NA)

# Calculando a profundidade da hierarquia após criar os níveis
df_powerbi["category_depth"] = df_powerbi[
    [f"category_level_{level}" for level in range(1, category_parts.shape[1] + 1)]
].notna().sum(axis=1).astype("Int64")

# Renomeando o caminho original e removendo identificadores de usuário e review
df_powerbi = df_powerbi.rename(columns={"category": "category_path"})
df_powerbi = df_powerbi.drop(columns=["user_id", "user_name", "review_id"])

# Exibindo informações sobre o DataFrame final para Power BI
print(f"Linhas preservadas: {len(df_powerbi):,} de {len(df):,}")
print(f"Colunas na saída analítica: {df_powerbi.shape[1]}")
print(f"product_id repetidos preservados: {df_powerbi['product_id'].duplicated().sum():,}")
print(f"Ratings nulos após conversão: {df_powerbi['rating'].isna().sum():,}")
print(f"rating_count nulos: {df_powerbi['rating_count'].isna().sum():,}")

# Exibindo os tipos de dados após o tratamento para Power BI
display(df_powerbi.dtypes.rename("tipo_destino").to_frame())
display(df_powerbi.head(3))
```

## 9. Validar e Exportar

As verificações de faixa ignoram valores nulos e mostram sua quantidade separadamente. Os CSVs são gravados em `data/processed/` com UTF-8 com BOM (Marca de Ordem dos Bytes) para facilitar a leitura no Power BI.

```python
# Verificando os dados não nulos antes da exportação
valid_price_rows = df_powerbi[["discounted_price", "actual_price"]].notna().all(axis=1)
price_rows = df_powerbi.loc[valid_price_rows]

# Verificando ocorrências fora dos critérios de qualidade
issue_counts = pd.Series(
    {
        "precos_promocionais_negativos": df_powerbi["discounted_price"].dropna().lt(0).sum(),
        "precos_originais_negativos": df_powerbi["actual_price"].dropna().lt(0).sum(),
        "precos_promocionais_maiores_que_originais": (
            price_rows["discounted_price"] > price_rows["actual_price"]
        ).sum(),
        "descontos_fora_de_zero_a_um": (~df_powerbi["discount_percentage"].dropna().between(0, 1)).sum(),
        "avaliacoes_fora_de_zero_a_cinco": (~df_powerbi["rating"].dropna().between(0, 5)).sum(),
        "contagens_de_avaliacoes_negativas": df_powerbi["rating_count"].dropna().lt(0).sum(),
    },
    name="quantidade_com_problema",
)

# Exibindo a quantidade de ocorrências fora dos critérios, considerando apenas valores não nulos
null_counts = df_powerbi[
    ["discounted_price", "actual_price", "discount_percentage", "rating", "rating_count"]
].isna().sum().rename("valores_nulos")

# Exibindo o resultado das verificações de qualidade
quality_checks = pd.Series(
    {
        "quantidade_de_linhas_preservada": len(df_powerbi) == len(df),
        "record_id_unico": df_powerbi["record_id"].is_unique,
        "precos_promocionais_nao_negativos": issue_counts["precos_promocionais_negativos"] == 0,
        "precos_originais_nao_negativos": issue_counts["precos_originais_negativos"] == 0,
        "precos_promocionais_nao_maiores_que_originais": issue_counts["precos_promocionais_maiores_que_originais"] == 0,
        "descontos_dentro_do_intervalo": issue_counts["descontos_fora_de_zero_a_um"] == 0,
        "avaliacoes_dentro_do_intervalo": issue_counts["avaliacoes_fora_de_zero_a_cinco"] == 0,
        "contagens_de_avaliacoes_nao_negativas": issue_counts["contagens_de_avaliacoes_negativas"] == 0,
        "identificadores_de_usuario_removidos": not any(
            column in df_powerbi.columns for column in ["user_id", "user_name", "review_id"]
        ),
    },
    name="resultado",
)

# Exibindo os resultados das verificações de qualidade
print("Ocorrências fora dos critérios, considerando apenas valores não nulos:")
display(issue_counts.to_frame())
print("Valores nulos nos campos numéricos:")
display(null_counts.to_frame())
print("Resultado das validações:")
display(quality_checks.to_frame())

# Verificando se todas as verificações de qualidade foram aprovadas
if not quality_checks.all():
    raise ValueError("Uma ou mais verificações falharam; revise os dados antes da exportação.")

# Exportando os arquivos tratados para a pasta processed
processed_dir = data_path.parent.parent / "processed"
processed_dir.mkdir(parents=True, exist_ok=True)
powerbi_path = processed_dir / "amazon_powerbi.csv"
dictionary_path = processed_dir / "dicionario_dados_amazon.csv"

# Exportando os DataFrames para CSV com codificação UTF-8 com BOM para compatibilidade com Power BI
df_powerbi.to_csv(powerbi_path, index=False, encoding="utf-8-sig")
data_dictionary.to_csv(dictionary_path, index=False, encoding="utf-8-sig")

print(f"\nBase tratada: {powerbi_path} ({df_powerbi.shape[0]:,} linhas x {df_powerbi.shape[1]} colunas)")
print(f"Dicionário: {dictionary_path} ({len(data_dictionary)} campos documentados)")
```

```python
# Recarregando os arquivos exportados para validação
powerbi_reloaded = pd.read_csv(powerbi_path)
dictionary_reloaded = pd.read_csv(dictionary_path)

print(f"CSV tratado relido: {powerbi_reloaded.shape[0]:,} linhas x {powerbi_reloaded.shape[1]} colunas")
print(f"Dicionário relido: {len(dictionary_reloaded):,} linhas")
print("Colunas numéricas após leitura do CSV:")
display(powerbi_reloaded.select_dtypes(include="number").dtypes.rename("tipo").to_frame())

# Confirmando dimensões, chave, documentação e codificação dos arquivos exportados
assert powerbi_reloaded.shape == df_powerbi.shape
assert len(powerbi_reloaded) == len(df)
assert powerbi_reloaded["record_id"].is_unique
assert "category_path" in powerbi_reloaded.columns
assert dictionary_reloaded["coluna"].eq("category_path").any()
assert powerbi_path.read_bytes().startswith(b"\xef\xbb\xbf")
assert dictionary_path.read_bytes().startswith(b"\xef\xbb\xbf")
```

### Configurações recomendadas para uso no Power BI

- Defina `product_id` como texto e `record_id` como número inteiro.
- Formate `discounted_price` e `actual_price` como moeda INR (₹), sem converter para reais.
- Formate `discount_percentage` como percentual; os valores já estão entre 0 e 1.
- Defina `rating_count` e `category_depth` como número inteiro; `rating_count` tem nulos e pode ser inferido como decimal no CSV.
- Mantenha `rating` como número decimal, os níveis de categoria como texto e os links como URL da Web.
- A tabela continua no grão dos registros originais. Para análises de reviews individuais, será necessária uma fonte estruturada, porque o texto agrupado não pode ser dividido com segurança.
